```python
plt.figure(figsize=(10, 5))
sns.histplot(df['price'].dropna(), bins=50, kde=True)
plt.title('Distribution of Car Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

<div dir="ltr">

### Chart 1 Analysis: Distribution of Car Prices

**Is the price distribution normal?**

No. The distribution is completely right-skewed. The highest concentration of data is in the range of $10,000 to $40,000,

and after that, it decreases rapidly. This is consistent with the mean of 40,326 and the median of 31,000 that we saw earlier.

**Are outliers observed?**

Yes. There is a long tail extending to around $300,000, indicating a small number of high-priced cars.

These cars are considered outliers.

</div>

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='model_year', y='price', alpha=0.4)
plt.title('Price vs Model Year')
plt.xlabel('Model Year')
plt.ylabel('Price')
plt.show()
```

<div dir="ltr">

### Chart 2 Analysis: Relationship Between Price and Manufacturing Year

**Is there a relationship between price and manufacturing year?**

Yes, there is a clear relationship. As we approach recent years, especially after the year 2000,

the spread and maximum prices increase noticeably. For cars manufactured before 1990,

we have almost no data, and only one point can be seen around the year 1974.

**Is this relationship positive or negative?**

It is positive. The newer the manufacturing year, the higher both the average price and the maximum possible prices

become. Clearly, from 2015 onward, points with prices above two hundred thousand dollars

appear, which were not seen at all in earlier years. This result is exactly consistent with the table

grouping the average price by year that we calculated earlier.

</div>

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='milage', y='price', alpha=0.4)
plt.title('Price vs Mileage')
plt.xlabel('Mileage')
plt.ylabel('Price')
plt.show()
```

<div dir="ltr">

### Chart 3 Analysis: Relationship Between Price and Mileage

**Does the price decrease as mileage increases?**

Yes, there is a very clear inverse relationship. The highest concentration of high prices, meaning above one hundred and

fifty thousand dollars, is located exactly near the zero-mileage axis. As mileage increases, both the density

of points and the maximum price decrease rapidly. The overall shape of the chart resembles a declining exponential curve.

**Are outliers observed?**

Yes. Two types of outliers can be observed in this chart. The first type consists of several cars with very high mileage,

around three hundred and fifty thousand to four hundred and five thousand miles, which are scattered on the right side of the chart.

The second type is a point with relatively low mileage, around one hundred and twenty thousand miles, but an unexpectedly high price,

around one hundred and five thousand dollars, which deviates from the overall pattern.

</div>

```python
plt.figure(figsize=(10, 5))
df.isnull().sum().sort_values(ascending=False).plot(kind='bar')
plt.title('Missing Values per Column')
plt.ylabel('Count of Missing Values')
plt.show()
```

<div dir="ltr">

### Chart 5 Analysis: Missing Values Assessment

**Which column has the most missing values?**

The document status column, meaning clean_title, has by far the highest number of missing values, with around five hundred and ninety-eight missing values.

After that, the fuel type column with around one hundred and seventy cases and the accident history column with around one hundred and fourteen cases are in the next positions.

The price column also has around twenty-eight missing values. The other columns, including brand, model, manufacturing year, and mileage, have no missing values.

**Is data cleaning needed before building the model?**

Yes, definitely. Given the high number of missing values in the document status column, which accounts for nearly fifteen percent of the total data,

this column should either be removed or filled using an appropriate method. Also, rows that do not have a value in the price column, since price is the target variable, should be removed before training the model.

The fuel type and accident history columns also need their missing values to be filled using a method such as replacing them with the most frequent value.
